Wichtigste Erkenntnisse (auf den Punkt)

1) ORSY-Kunden sind deutlich aktiver & wertvoller
	•	Höhere Bestellfrequenz (orders_count), breiteres Sortiment (diff_products_count) und höherer Umsatz (sales) als Nicht-ORSY.
	•	Interpretation: Stärkere Bindung, regelmäßige Nachbestellungen, höhere Kundenlebenszeit.

2) Digitalisierung treibt ORSY
	•	ORSY-Käufer haben höheren Digitalisierungs-Score (digital_flags_share) und mehr E-Shop-Logins.
	•	Hebel: App/E-Shop/eProcurement pushen → höhere ORSY-Adoption.

3) Regionen
	•	Region 16: höchste ORSY-Quote (13,9 %).
	•	Regionen 11–15/17–18: ~11–12 %.
	•	Ansatz: Best Practices aus Region 16 identifizieren und ausrollen.

4) Marktsegmente
	•	Segment 4: 31,3 % ORSY-Quote und höchster Umsatz/Ku. (≈ 9 091 €) → Premium-Zielgruppe.
	•	Segmente 13/17/18: ~16–17 % ORSY-Quote → gute Ausbaukandidaten.

5) Unternehmensgröße
	•	Mittelgroße Firmen (6–50 MA) zeigen höhere ORSY-Affinität.
	•	Sehr kleine (0–1) und sehr große (>200) nutzen ORSY seltener.
	•	Ansatz: Mid-Market fokussieren, für Enterprise evtl. andere Angebote/Prozesse.
